```python
import os 
from dotenv import load_dotenv
load_dotenv() ## aloading all the environment veriable

groq_api_key=os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
model=ChatGroq(model="llama-3.1-8b-instant",groq_api_key=groq_api_key)
model.invoke([HumanMessage(content="Hi, My is ashish and i am a solution engg")])
```

```python
from langchain_core.messages import AIMessage
model.invoke([
    HumanMessage(content="Hi My name ashish, I am a solutions engineer"),
                 AIMessage(content="Nice to meet you, Ashish.  As a solution engineer, I'm sure you deal with a wide range of technical challenges and work closely with customers or stakeholders to understand their needs and propose suitable solutions. What kind of solutions do you typically work on?"),
                 HumanMessage(content="Hey What's my name and what do I do?")
])
```

### Message History

##### We can use a message history class to wrap our model and make it stateful. This will keep track of inputs and outputs of the model, and store them in datastore. Future interactions will then load those messages and pass them into the chain as part of the input. Let's see how to use this!

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store={}

def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history=RunnableWithMessageHistory(model,get_session_history)
```

```python
config={"configurable":{"session_id":"chat-2"}}
```

```python
with_message_history.invoke(
    [HumanMessage(content="Hi my name is ashish and i am a solutions engineer")],
    config=config
)
```

```python
with_message_history.invoke(
    [HumanMessage(content="whats my name")],
    config=config
)
```

```python
with_message_history.invoke(
    [HumanMessage(content="my name is pintu")],
    config=config
)
```

```python
with_message_history.invoke(
    [HumanMessage(content="whats my name?")],
    config=config
)
```
